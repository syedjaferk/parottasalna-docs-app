## Dictionary

```python
profile = {
    "name": "Syed Jafer K",
    "age": 28
}
```

```python
print(profile)
```

```python
print(profile["name"])
```

```python
print(profile["age"])
```

```python
profile = {
    "name": "Syed Jafer K",
    "age": 28,
    "marks": {
        "tamil": 95,
        "maths": 45,
        "biology": 75
    }
}
```

```python
print(profile)
```

```python
print(profile["marks"])
```

```python
print(profile["marks"]["tamil"])
```

```python
profile["marks"]["tamil"] = 85
```

```python
print(profile)
```

```python
profile["marks"]["social"] = 45
```

```python
print(profile)
```

```python
print(profile["address"])
```

```python
print(profile.get("address"))
```

```python
print(profile.get("address", 123))
```

```python
dir(profile)
```

```python
profile.keys()
```

```python
profile.values()
```

```python
for key, value in profile.items():
  print(key, "-----", value)
```

```python
students_profile = [profile]
```

```python
parthiban = {'name': 'Parthiban', 'age': 28, 'marks': {'tamil': 95, 'maths': 89, 'biology': 75, 'social': 45}}
```

```python
students_profile.append(parthiban)
```

```python
print(students_profile)
```

```python
for student in students_profile:
  print(student["name"], student["marks"]["tamil"])
```
